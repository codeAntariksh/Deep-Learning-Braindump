## Removing Punctuations

```python
import string  # For punctuation, whitespace, etc.
import time    # For handling time-related tasks

# Now you can use them:
print(string.punctuation)
print(time.ctime())
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char,'')
    return text
```

```python
text = "string. With. Punctuation?"
```

```python
start = time.time()
print(remove_punc(text))
time1 = time.time() - start
print(time1*50000)
```

```python
def remove_punc1(text):
  return text.translate(str.maketrans('','',exclude))
```

```python
slang_dict = {
    line.split('=')[0].strip(): line.split('=')[1].strip()
    for line in DB.strip().splitlines()
    if '=' in line
}
```

```python
def translator(user_string):
    # Standardize whitespace and split into words
    words = user_string.split()
    new_words = []

    for word in words:
        # Remove punctuation to match the dictionary key
        # e.g., "ASAP!" becomes "ASAP"
        clean_word = re.sub(r'[^\w]', '', word).upper()

        # Look up in dictionary, keep original if not found
        translated = slang_dict.get(clean_word, word)
        new_words.append(translated)

    return ' '.join(new_words)
```

## Removing StopWords

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
stopwords.words("english")
```

```python
def remove_stopwords(text):
  new_text = []
  for word in text.split():
    if word in stopwords.words("english"):
      new_text.append('')
    else:
      new_text.append(word)
  x = new_text[:]
  new_text.clear()
  return " ".join(x)
```

```python
df["review"].apply(remove_stopwords)
```
