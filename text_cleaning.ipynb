## HTML Tag Cleaning

```python
sample_text = """<section class="hero-section">
                <h1 class="hero-heading"><span class="highlight highlight-secondary" >Mindful living</span> for the<br class="hero-heading-break"> digital world📱</h1>
                <p class="hero-paragraph">The ultimate bullet journal app to document, reflect and embrace your personal journey.</p>
                <ul class="hero-nav-list">
                    <li><a href="register.html" class="btn btn-primary">Get Started</a></li>
                    <li><a href="#features" class="btn btn-primary-outline">Learn More</a></li>
                </ul>
            </section>"""
```

```python
import re
def striphtml(data):
  p = re.compile(r'<.*?>')
  return p.sub('',data)
```

```python
striphtml(sample_text)
```

```python
example = "<p>Hii, How are You Doing Jake?</p><br><br><p> I am Doing Fine Grim</p>"
```

```python
striphtml(example)
```

## Tokenization

```python
para = "The art of writing has evolved from ink on parchment to pixels on a glowing screen, yet the core intent remains unchanged: to connect one mind to another. In our fast-paced digital era, the ability to slow down and craft a meaningful sentence is a form of quiet rebellion against the noise. We often find ourselves caught between the precision of algorithms and the messy, beautiful spontaneity of human thought. However, it is exactly in that middle ground—where logic meets imagination—that the most impactful stories are born. Whether we are coding a new future or simply texting a friend, our words serve as the bridge that turns isolation into community."
```

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
tokens = word_tokenize(para)
```

```python
print(tokens)
```

```python
from nltk.tokenize import sent_tokenize
token = sent_tokenize(para)
```

```python
for sent in token:
  print(word_tokenize(sent))
```
